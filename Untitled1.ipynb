importing the libraries

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("/content/credit_default_classification.csv")

```

```python
df.info()
```

```python
df.isnull().sum()
```

handling the missing values

```python

mean_columns = ["loan_amount_k","credit_score","income_k","months_since_delinq","credit_score_dup"]
df[mean_columns] = df[mean_columns].fillna(df[mean_columns].mean())

df["age"] = df["age"].fillna(df["age"].median())

df["num_open_accounts"] = df["num_open_accounts"].fillna(df["num_open_accounts"].median())

df["home_owner"] = df["home_owner"].fillna(df["home_owner"].mode()[0])
df["loan_purpose"] = df["loan_purpose"].fillna(df["loan_purpose"].mode()[0])
df["defaulted"].ffill()

df["customer"] = np.arange(1,258)

```

handles the inconsistent data

```python
df["loan_purpose"].replace(["Debt cnsolidation",'Debt consolidtion',"Debt consolidation","Debt consolidition","Dbt consolidation","Debt consolidaton","DEBT CONSOLIDATION"],"Debt Consolidation",inplace = True)
df["loan_purpose"].replace(["Home improvemnt","Home improvement","Home imprvement","HOME IMPROVEMENT","Home imrovement","Hme improvement","Hoe improvement","ome improvement","Hom improvement"],"Home Improvement",inplace = True)
df["loan_purpose"].replace(["Business","Busines","BUSINESS","Buiness","usiness","Businss","Busness","Busiess","Bsiness"],"Business",inplace = True)
df["loan_purpose"].replace(["Cr","CAR","ar"],"Car",inplace = True)
df["loan_purpose"].replace(["EDUCATION","Educatin","ducation","Eduction","Educaion","Educatio","Edcation","Eucation"],"Education",inplace = True)

df["home_owner"].replace(["Ye","Ys","es","YES"],"Yes",inplace = True)
df["home_owner"].replace(["NO","N",'o'],"No",inplace = True)

df["income_k"] = df["income_k"].abs().round(2)
df["credit_score"] = df["credit_score"].round(0)
```

Now, we'll analyze relationships between two variables, particularly focusing on how features relate to loan defaults.

# Credit Default Classification Dataset Analysis Report

## 1. Executive Summary

The Credit Default Classification dataset contains **257 records and 10 columns**. It includes information about borrowers, such as their income, credit score, loan amount, age, homeownership, and loan purpose.

The main aim of this analysis is to understand the factors that may influence whether a borrower defaults on a loan. By exploring the dataset through different visualizations and statistical methods, we can identify patterns and relationships between borrowers' financial details and their default status.

## 2. Data Quality & Preprocessing Assessment

Before starting the analysis, the dataset was checked and cleaned to make sure the information was consistent and suitable for further exploration.

The following preprocessing steps were performed:

- **Handling Missing Values:** Missing values in columns such as `age`, `loan_amount_k`, and `credit_score` were filled using appropriate methods like mean, median, and mode, depending on the type of data.
- **Removing Duplicate Columns:** The unnecessary `credit_score_dup` column was removed because it contained duplicate information and was not needed for the analysis.
- **Cleaning Customer IDs:** The inconsistent values in the `customer` column were replaced with unique and properly formatted IDs ranging from `CR30000` to `CR30256`.
- **Correcting Categorical Values:** Different spellings and formats of loan purposes, such as 'Debt cnsolidation', 'Debt consolidtion', and 'DEBT CONSOLIDATION', were identified and combined into consistent categories. The final categories include Debt Consolidation, Home Improvement, Business, Car, and Education.

These steps helped improve the overall quality of the dataset and made it easier to perform further analysis.

## 3. Univariate Analysis (Understanding Individual Features)

Univariate analysis was performed to understand the distribution of individual columns in the dataset.

- **Loan Purpose:** Business loans were the most common, with 77 records. Home Improvement accounted for 48 loans, followed by Debt Consolidation with 46, Car loans with 44, and Education loans with 42. The distribution shows that borrowers have different reasons for taking loans, with Business being the most popular purpose.
- **Home Ownership:** Around 57.6% of borrowers do not own a house, while the remaining 42.4% are homeowners. This shows that a slightly higher number of borrowers in the dataset do not own their homes.

## 4. Bivariate & Multivariate Analysis

Different visualizations were used to understand how two or more features are related to each other.

- **Credit Score vs. Default Status:** The box plot shows that the median credit score is approximately 565. However, the visualization displays only one class (0.0) for the `defaulted` column. This could be due to class imbalance or because missing values in the target column were filled using the mode, which may have affected the original distribution.
  
- **Income vs. Loan Amount:** The scatter plot and bubble chart show how loan amounts vary with borrower income. Loan amounts reach up to approximately $500K, while incomes go up to around $130K. The points are spread across the chart without showing a clear linear relationship. This suggests that income alone may not explain how much money a borrower chooses to borrow.

- **Correlation Analysis:** The heatmap shows that the numerical features have very weak linear relationships with each other, with correlation values ranging from approximately -0.06 to +0.13. This indicates that there are no strong linear relationships among the numerical variables. Therefore, models that can identify more complex patterns, such as Random Forest and Gradient Boosting, may be worth exploring.
